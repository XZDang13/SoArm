```python
from PIL import Image
```

```python
L = Image.open("sim.png").convert("RGBA")
R = Image.open("800_Color.png").convert("RGBA")
```

```python
total_w = L.width + R.width
split_x = int(total_w * 0.5)
```

```python
left_w = min(split_x, L.width)
right_w = max(0, (L.width + R.width) - split_x)

# Crop left part from L and right part from R
left_part  = L.crop((0, 0, left_w, L.height))
# For right part, we want the rightmost 'right_w' pixels of R
right_part = R.crop((max(0, R.width - right_w), 0, R.width, R.height))
```

```python
H = max(left_part.height, right_part.height)
canvas = Image.new("RGBA", (left_part.width + right_part.width, H), (0, 0, 0, 0))

# Paste hard seam by default
canvas.paste(left_part, (0, 0))
canvas.paste(right_part, (left_part.width, 0))
```
