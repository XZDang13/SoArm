```python
from PIL import Image
```

```python
L = Image.open("sim.png")
R = Image.open("real_640_480.png")
```

```python
L.size
```

```python
R.size
```

```python
half_L = L.width // 2
half_R = R.width // 2

left_part = L.crop((0, 0, half_L, L.height))
right_part = R.crop((0, 0, half_L, R.height))

out = Image.new("RGB", (half_L + half_R, L.height))
out.paste(left_part, (0, 0))
out.paste(right_part, (half_L, 0))
```
