```python
import time
import torch
from PIL import Image
from torchvision.transforms import v2
from model.actor_critic import FrameObservationEncoderNet, MobileFrameObservationEncoderNet
```

```python
device = torch.device("cuda:0")
frame_encoder = FrameObservationEncoderNet(128).to(device)
weight, _, _ = torch.load("frame_model.pth", weights_only=True)
frame_encoder.load_state_dict(weight)
frame_encoder = frame_encoder.eval()
```

```python
real_img = Image.open("debug_data/real_0.png")
sim_img = Image.open("sim.png")
```

```python
transform = v2.Compose([
            v2.ToImage(),
            v2.Resize((112, 112)),
            v2.ToDtype(torch.float32, scale=True),
            v2.Normalize((0.485, 0.456, 0.406), (0.229, 0.224, 0.225)),
        ])
```

```python
real_input = transform(real_img).to(device)
sim_input = transform(sim_img).to(device)
```

```python
inputs = torch.stack([real_input, sim_input], dim=0)
```

```python
outputs = frame_encoder(inputs)
```

```python
torch.cosine_similarity(outputs[0], outputs[1], dim=0)
```
