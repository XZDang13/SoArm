```python
import time
import torch
from PIL import Image
from torchvision.transforms import v2
```

```python
net = torch.hub.load('facebookresearch/dinov2', 'dinov2_vits14')
```

```python
device = torch.device("cuda:0")
net = net.to(device)
```

```python
real_img = Image.open("imgs/0_cur.png").convert("RGB")
sim_img = Image.open("sim.png").resize((640, 360))
```

```python
transform = v2.Compose([
            v2.ToImage(),
            v2.CenterCrop((360, 360)),
            v2.Resize((224, 224)),
            v2.ToDtype(torch.float32, scale=True),
            v2.Normalize((0.485, 0.456, 0.406), (0.229, 0.224, 0.225)),
        ])
```

```python
real_input = transform(real_img).unsqueeze(0).to(device)
sim_input = transform(sim_img).unsqueeze(0).to(device)
```

```python
real_f = net(real_input)
sim_f = net(sim_input)
```

```python
torch.cosine_similarity(real_f, sim_f)
```
